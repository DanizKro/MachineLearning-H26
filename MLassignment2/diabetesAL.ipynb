```python
import numpy as np
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib
import sklearn

# Set the directory in which to store data
NB_DIR = Path.cwd()       # Set NB_DIR to be the current working directory
DATA = NB_DIR/'data'      # The data dir is the subdirectory 'data' under NB_DIR

DATA.mkdir(exist_ok=True) # Create the data directory
```

```python
import urllib.request

# Define the file path
FILE = DATA / 'train.csv'
```

```python
diabetes = pd.read_csv(FILE)
```

```python
diabetes.head()
```

```python
len(diabetes)
```

```python
from sklearn.model_selection import train_test_split

# Først: sett av 10 % til test fra "diabetes"
diabetes_sett, diabetes_test = train_test_split(diabetes, test_size=0.10, random_state=42)

# Deretter: sett av 10 % av hele "diabetes_sett" til verifisering
# 10 % av hele datasettet er 1/9 av de resterende 90 %
diabetes_train, diabetes_verify = train_test_split(diabetes_sett, test_size=1/9, random_state=42)
```

```python
# Inputvariabler: alle kolonner unntatt target
X_train = diabetes_train.drop(columns=["Target", "Id"])
y_train = diabetes_train["Target"]

X_verify = diabetes_verify.drop(columns=["Target", "Id"])
y_verify = diabetes_verify["Target"]

X_test_holdout = diabetes_test.drop(columns=["Target", "Id"])
y_test_holdout = diabetes_test["Target"]
```

```python
print(len(X_train))
print(len(X_verify))
print(len(X_test_holdout))
```

```python
from sklearn.ensemble import RandomForestClassifier

#Modellen
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced"
)

model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
#Modellen lager prediksjoner for verifiseringsdataene
y_verify_pred = model.predict(X_verify)

print("Accuracy:", accuracy_score(y_verify, y_verify_pred))
print("\nClassification report:")
print(classification_report(y_verify, y_verify_pred))
print("\nConfusion matrix:")
print(confusion_matrix(y_verify, y_verify_pred))
```

```python
y_test_holdout_pred = model.predict(X_test_holdout)

print("Holdout accuracy:", accuracy_score(y_test_holdout, y_test_holdout_pred))
print("\nHoldout classification report:")
print(classification_report(y_test_holdout, y_test_holdout_pred))
print("\nHoldout confusion matrix:")
print(confusion_matrix(y_test_holdout, y_test_holdout_pred))
```

```python
import pandas as pd

test_csv = pd.read_csv(DATA/"test.csv")

X_test_csv = test_csv.drop(columns="Id")
X_test_csv = X_test_csv[X_train.columns]

y_test_csv_pred = model.predict(X_test_csv)

resultat = test_csv[["Id"]].copy()
resultat["Target"] = y_test_csv_pred

resultat.to_csv("prediksjoner.csv", index=False)
```

```python
print(resultat.head())
print(resultat.shape)
```

```python
#Hvordan blir variablene vektlagt? sjekker
import pandas as pd

viktighet = pd.Series(
    model.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

print(viktighet)
```

```python
#Sjekker hva som skjer med ferre variabler


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

rangert = viktighet.index.tolist()

for antall in [len(rangert), 15, 10]:
    kolonner = rangert[:antall]

    modell = RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        class_weight="balanced"
    )

    modell.fit(X_train[kolonner], y_train)
    prediksjon = modell.predict(X_verify[kolonner])

    print(f"\nAntall variabler: {antall}")
    print("Accuracy:", accuracy_score(y_verify, prediksjon))
    print(classification_report(y_verify, prediksjon))
```

```python
#Sjekker forskjellei modeller

from sklearn.metrics import classification_report

# Bruk modellen som er trent med alle 21 variablene
sannsynlighet_1 = model.predict_proba(X_verify)[:, 1]

for grense in [0.5, 0.4, 0.3, 0.2]:
    prediksjon = (sannsynlighet_1 >= grense).astype(int)

    print(f"\nBeslutningsgrense: {grense}")
    print(classification_report(y_verify, prediksjon))
```

```python
#Last inn Kaggle-testsettet
kaggle_test = pd.read_csv(DATA / "test.csv")
```

```python
# kaggle_test er lest inn fra test.csv
X_kaggle_test = kaggle_test.drop(columns="Id")
X_kaggle_test = X_kaggle_test[X_train.columns]

sannsynlighet_test = model.predict_proba(X_kaggle_test)[:, 1]
prediksjon_test = (sannsynlighet_test >= 0.5).astype(int)

submission = kaggle_test[["Id"]].copy()
submission["Target"] = prediksjon_test

submission.to_csv("submission.csv", index=False)
```
